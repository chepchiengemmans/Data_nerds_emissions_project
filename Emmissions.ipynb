## Business Understanding

### Background
Climate change and the global transition toward cleaner energy have increased pressure on governments, businesses, and investors to reduce carbon dioxide (CO₂) emissions. However, countries differ significantly in their emissions levels, economic activity, population, energy consumption, carbon intensity, and dependence on fossil fuels.

These differences make it difficult to apply the same clean-energy strategy across all countries. Grouping countries according to similar environmental, economic, demographic, and energy characteristics can help decision-makers better understand different country profiles and identify where particular clean-energy solutions and interventions may be most relevant.
Using historical country-level data on CO₂ emissions, GDP, population, energy consumption, and emission sources provides an opportunity to identify meaningful patterns and similarities among countries through data-driven analysis.

### Business Problem
Businesses, investors, and policymakers face challenges in identifying and comparing countries with similar clean-energy needs. Relying only on total CO₂ emissions may overlook important differences in economic activity, population, energy consumption, carbon intensity, and sources of emissions.
Without a systematic way of grouping countries according to these characteristics, decision-makers may struggle to identify markets with similar environmental challenges and determine which groups of countries should receive greater attention for clean-energy solutions and emission-reduction interventions.

### Business Objective
The objective of this project is to use country-level CO₂ emissions, economic, demographic, and energy data to identify groups of countries with similar characteristics.
By developing country profiles through clustering, the project aims to help businesses, investors, policymakers, and development organizations better understand differences between markets and identify groups that may require different clean-energy strategies or levels of intervention.

## Data Preprocessing

### Steps in the data preprocessing pipeline
1. Remove years below 1950.
2. Removing columns with more than 50% missing.
3. Impute missing values.
4. Removing outliers.
5. Separate continents from countries in their own datasets.

```python
# Preprocessing pipeline
def preprocessing_pipeline(data):
    # Making a copy of the original dataset
    data = data.copy()

    # 1. Removing years below 1950
    data = data[data['year'] >= 1950]

    # 2. Removing columns with more than 50% missing
    data = data.loc[:, data.isnull().mean() <= 0.5]

    # Replace all values of 0 with NaN
    data = data.replace(0, np.nan)
    
    # 3. Imputing missing values
    # 3.1. Get the descriptive statistics dataframe
    desc = data.describe()

    # 3.2. Isolate the rows for Q1, Q2, and Q3
    quartiles = desc.loc[["25%", "50%", "75%"]]

    # 3.3. Find which row name ("25%", "50%", or "75%") holds the largest value per column
    largest_quartile = quartiles.idxmax(axis=0)
    
    # 3.4. Loop through columns and apply the corresponding imputation
    for col in df.columns:
        # Skip columns that aren't numeric (not present in df.describe)
        if col not in largest_quartile.index:
            continue

        # Rule: Use mean if Q2 ("50%") is the largest
        if largest_quartile[col] == "50%":
            fill_value = desc.loc["mean", col]

        # Rule: Use median if Q1 ("25%") or Q3 ("75%") is the largest
        else:
            fill_value = desc.loc["50%", col]

        # Apply the imputation to the column
        data[col] = data[col].fillna(fill_value)

    # 5. Separating continents from countries in their own datasets.
    # List of all continents
    continent = [
        'Africa',
        'Africa (GCP)'
        'Asia',
        'Asia (GCP)',
        'Asia (excl. China and India)',
        'Europe',
        'Europe (GCP)',
        'Europe (excl. EU-27)',
        'Europe (excl. EU-28)',
        'European Union (27)',
        'European Union (28)',
        'North America',
        'Central America (GCP)',
        'North America',
        'North America (GCP)',
        'North America (excl. USA)',
        'South America',
        'South America (GCP)',
        'Oceania',
        'Oceania (GCP)'
        ]

    # Actually separating the continents
    data_continent = data[data['country'].isin(continent)]

    # Returning the dataset
    return data, data_continent

df_cleaned, df_continent = preprocessing_pipeline(df)
df_cleaned.isnull().sum()
#df_cleaned.head(50)
```
